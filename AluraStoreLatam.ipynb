### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
print(tienda.columns.tolist())
```

#1. Análisis de facturación



```python
facturacion_tienda1 = tienda["Precio"].sum()
facturacion_tienda2 = tienda2["Precio"].sum()
facturacion_tienda3 = tienda3["Precio"].sum()
facturacion_tienda4 = tienda4["Precio"].sum()
```

```python
print("FACTURACIÓN TOTAL POR TIENDA")

print(f"Tienda 1: ${facturacion_tienda1:,.2f}")
print(f"Tienda 2: ${facturacion_tienda2:,.2f}")
print(f"Tienda 3: ${facturacion_tienda3:,.2f}")
print(f"Tienda 4: ${facturacion_tienda4:,.2f}")
```

# 2. Ventas por categoría

```python
print("CATEGORÍAS MÁS POPULARES POR TIENDA")


for nombre, df in [("Tienda 1", tienda), ("Tienda 2", tienda2),
                   ("Tienda 3", tienda3), ("Tienda 4", tienda4)]:
    print(f"\n {nombre}:")
    print(df["Categoría del Producto"].value_counts())
```

# 3. Calificación promedio de la tienda


```python
promedio1 = tienda["Calificación"].mean()
promedio2 = tienda2["Calificación"].mean()
promedio3 = tienda3["Calificación"].mean()
promedio4 = tienda4["Calificación"].mean()
```

```python
print("CALIFICACIÓN PROMEDIO POR TIENDA")

print(f"Tienda 1: {promedio1:.2f} / 5")
print(f"Tienda 2: {promedio2:.2f} / 5")
print(f"Tienda 3: {promedio3:.2f} / 5")
print(f"Tienda 4: {promedio4:.2f} / 5")
```

# 4. Productos más y menos vendidos

```python
def productos_extremos(df, nombre):
    conteo = df["Producto"].value_counts()
    print(f"\n {nombre}")
    print(f"  Más vendido:   {conteo.index[0]}  ({conteo.iloc[0]} veces)")
    print(f"  Menos vendido: {conteo.index[-1]}  ({conteo.iloc[-1]} veces)")

print("PRODUCTOS MÁS Y MENOS VENDIDOS")
productos_extremos(tienda,  "Tienda 1")
productos_extremos(tienda2, "Tienda 2")
productos_extremos(tienda3, "Tienda 3")
productos_extremos(tienda4, "Tienda 4")
```

# 5. Envío promedio por tienda

```python
envio1 = tienda["Costo de envío"].mean()
envio2 = tienda2["Costo de envío"].mean()
envio3 = tienda3["Costo de envío"].mean()
envio4 = tienda4["Costo de envío"].mean()

```

```python
print("COSTO DE ENVÍO PROMEDIO POR TIENDA")
print(f"Tienda 1: ${envio1:,.2f}")
print(f"Tienda 2: ${envio2:,.2f}")
print(f"Tienda 3: ${envio3:,.2f}")
print(f"Tienda 4: ${envio4:,.2f}")
```

# 1. Facturación total por tienda

```python
import matplotlib.pyplot as plt

tiendas     = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
facturaciones = [facturacion_tienda1, facturacion_tienda2,
                 facturacion_tienda3, facturacion_tienda4]
colores     = ["gold", "skyblue", "limegreen", "hotpink"]

fig, ax = plt.subplots(figsize=(9, 5))
barras = ax.bar(tiendas, facturaciones, color=colores, edgecolor="white", width=0.5)

for barra in barras:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2, altura + 50000,
            f"${altura:,.0f}", ha="center", fontsize=9, fontweight="bold")

ax.set_title("Facturación Total por Tienda", fontsize=14, fontweight="bold", pad=15)
ax.set_xlabel("Tienda", fontsize=11)
ax.set_ylabel("Facturación ($)", fontsize=11)
ax.set_ylim(0, max(facturaciones) * 1.15)
plt.tight_layout()
plt.show()
```

El presente informe tiene como objetivo analizar el rendimiento de las cuatro
tiendas pertenecientes al Sr. Juan, con el fin de identificar cuál de ellas

Para llegar a esta conclusión se evaluaron cinco dimensiones clave:
facturación total, categorías de productos más vendidas, calificación promedio
de clientes, productos más y menos vendidos, y costo de envío promedio.


## 2. Análisis



La *Tienda 1* lidera en ingresos, mientras que la *Tienda 4* es la que
menos factura, con una diferencia de **$112,504,700** respecto a la líder,
lo que representa un 9.8% menos de ingresos. Este es el indicador más
crítico del análisis, ya que refleja directamente la capacidad de cada
tienda para generar valor económico.


###  Categorías de Productos

Las cuatro tiendas comparten el mismo patrón de ventas por categoría:
**Muebles y Electrónicos** lideran consistentemente, mientras que
**Artículos para el hogar e Instrumentos musicales** son las menos vendidas
en todas las tiendas.

Esta uniformidad indica que **no existe una ventaja competitiva diferencial**
entre tiendas basada en el portafolio de productos. Sin embargo, se destaca
que la Tienda 4 registra el menor volumen en Electrodomésticos (254 unidades)
e Instrumentos musicales (170 unidades), las dos categorías más bajas de todo
el grupo.



### Calificación Promedio de Clientes

| Tienda   | Calificación |
|----------|-------------|
| Tienda 1 | 3.98 / 5 ⚠️ |
| Tienda 2 | 4.04 / 5    |
| Tienda 3 | 4.05 / 5 🥇 |
| Tienda 4 | 4.00 / 5    |

Las diferencias en calificación son pequeñas entre tiendas, pero la
**Tienda 1 tiene la peor calificación promedio (3.98)**, siendo la única
por debajo de 4.00. Esto sugiere una leve insatisfacción relativa de sus
clientes frente a las demás tiendas.


### roductos Más y Menos Vendidos

| Tienda   | Más vendido                  | Menos vendido              |
|----------|------------------------------|----------------------------|
| Tienda 1 | Microondas (60 veces)        | Celular ABXY (33 veces)    |
| Tienda 2 | Iniciando en programación (65)| Juego de mesa (32 veces)  |
| Tienda 3 | Kit de bancas (57 veces)     | Bloques de construcción (35)|
| Tienda 4 | Cama box (62 veces)          | Guitarra eléctrica (33 veces)|

La **Tienda 3** presenta el producto más vendido con menor frecuencia
(57 veces), lo que indica que ningún producto logra destacarse con fuerza
en esa tienda. La **Tienda 2** tiene el producto más vendido con mayor
rotación (65 veces), lo que refleja una demanda más concentrada y predecible.



### Costo de Envío Promedio



La **Tienda 4 tiene el costo de envío más bajo**, lo que podría ser una
ventaja competitiva para atraer clientes. Por el contrario, la **Tienda 1
tiene el costo de envío más alto**, lo que puede ser un factor que
desincentiva compras y explica parcialmente su menor calificación.



## 3. Conclusión y Recomendación

###

## Se recomienda vender la Tienda 4.

A pesar de tener el costo de envío más bajo (una ventaja logística),
la **Tienda 4 acumula las mayores debilidades en los indicadores más
importantes**:

-  **Menor facturación total** del grupo: $1,038,375,700,  siendo
  $112,504,700 menos que la Tienda 1
-  **Menor volumen en categorías clave** como Electrodomésticos
  (254 unidades) e Instrumentos musicales (170 unidades).
-  El bajo costo de envío no se traduce en mayores ingresos,
  lo que indica que la tienda no está aprovechando esa ventaja para
  atraer más clientes o cerrar más ventas.

El costo de envío bajo es la única métrica favorable de la Tienda 4,
pero no compensa su rendimiento comercial significativamente inferior.
En contraste, las Tiendas 1, 2 y 3 presentan facturaciones notablemente
superiores y mayor volumen de ventas en categorías estratégicas.

Por lo tanto, la Tienda 4 es la candidata más lógica para ser vendida
permitiéndole al Sr. Juan concentrar sus recursos y esfuerzos en las
tres tiendas con mejor desempeño comprobado.
